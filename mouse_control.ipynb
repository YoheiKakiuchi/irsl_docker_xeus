```python
class HD(object):
    def __init__(self):
        self.dragging = False
        self.prev_x = -1
        self.prev_y = -1
    def mouse_move(self, x, y):
        if self.dragging:
            #print("Mouse move event:", x, y)
            cds, fov = ib.getCameraCoords()
            xx = x - self.prev_x
            yy = y - self.prev_y
            if xx != 0:
                cds.rotate(0.008*xx, coordinates.Y)
                cds.rotate(-0.008*yy, coordinates.X)
                ib.setCameraCoords(cds, fov)
                img = di.getImage()
                canvas.put_image_data(img)
            self.prev_x = x
            self.prev_y = y
    def mouse_down(self, x, y):
        self.dragging = True
        self.prev_x = x
        self.prev_y = y
    def mouse_up(self, x, y):
        self.dragging = False
        self.prev_x = -1
        self.prev_y = -1
```

```python
obj = HD()
```

```python
canvas.on_mouse_down(obj.mouse_down)
```

```python
canvas.on_mouse_up(obj.mouse_up)
```

```python
canvas.on_mouse_move(obj.mouse_move)
```

```python
canvas.on_mouse_move(obj.mouse_move, True)
canvas.on_mouse_up(obj.mouse_up, True)
canvas.on_mouse_down(obj.mouse_down, True)
```
